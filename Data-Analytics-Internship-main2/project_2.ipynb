```python
# ============================================================
# PROJECT 2: DATA ANALYSIS
# Goal: Analyze patterns, trends, and distributions
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files


# ------------------------------------------------------------
# 1. Load the Cleaned Dataset
# ------------------------------------------------------------

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_excel(filename)

print("Dataset loaded successfully.")
print("Dataset Shape:", df.shape)

print("\nFirst 5 rows:")
print(df.head())


# ------------------------------------------------------------
# 2. Basic Statistical Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("BASIC STATISTICS")
print("=" * 60)

numerical_columns = [
    "Quantity",
    "UnitPrice",
    "ItemsInCart",
    "TotalPrice"
]

print("\nDescriptive Statistics:")
print(df[numerical_columns].describe())

print("\nMean:")
print(df[numerical_columns].mean())

print("\nMedian:")
print(df[numerical_columns].median())


# ------------------------------------------------------------
# 3. Product Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PRODUCT ANALYSIS")
print("=" * 60)

print("\nNumber of Orders by Product:")
print(df["Product"].value_counts())

print("\nTotal Quantity Sold by Product:")
print(
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("\nRevenue by Product:")
print(
    df.groupby("Product")["TotalPrice"]
    .sum()
    .sort_values(ascending=False)
)


# ------------------------------------------------------------
# 4. Order Status Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ORDER STATUS ANALYSIS")
print("=" * 60)

print("\nOrder Status:")
print(df["OrderStatus"].value_counts())

print("\nOrder Status Percentage:")
print(
    (df["OrderStatus"].value_counts(normalize=True) * 100).round(2)
)


# ------------------------------------------------------------
# 5. Payment Method Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("PAYMENT METHOD ANALYSIS")
print("=" * 60)

print("\nPayment Methods:")
print(df["PaymentMethod"].value_counts())

print("\nPayment Method Percentage:")
print(
    (df["PaymentMethod"].value_counts(normalize=True) * 100).round(2)
)


# ------------------------------------------------------------
# 6. Referral Source Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("REFERRAL SOURCE ANALYSIS")
print("=" * 60)

print("\nOrders by Referral Source:")
print(df["ReferralSource"].value_counts())

print("\nRevenue by Referral Source:")
print(
    df.groupby("ReferralSource")["TotalPrice"]
    .sum()
    .sort_values(ascending=False)
)


# ------------------------------------------------------------
# 7. Monthly Revenue Trend
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MONTHLY REVENUE TREND")
print("=" * 60)

df["Month"] = df["Date"].dt.to_period("M")

monthly_revenue = df.groupby("Month")["TotalPrice"].sum()

print("\nMonthly Revenue:")
print(monthly_revenue)

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_revenue.index.astype(str),
    monthly_revenue.values,
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 8. Monthly Order Trend
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("MONTHLY ORDER TREND")
print("=" * 60)

monthly_orders = df.groupby("Month").size()

print("\nMonthly Number of Orders:")
print(monthly_orders)

plt.figure(figsize=(10, 5))

plt.plot(
    monthly_orders.index.astype(str),
    monthly_orders.values,
    marker="o"
)

plt.title("Monthly Order Trend")
plt.xlabel("Month")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 9. Total Price Distribution
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("TOTAL PRICE DISTRIBUTION")
print("=" * 60)

plt.figure(figsize=(8, 5))

plt.hist(
    df["TotalPrice"],
    bins=20,
    edgecolor="black"
)

plt.title("Distribution of Total Price")
plt.xlabel("Total Price")
plt.ylabel("Number of Orders")
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 10. Quantity Distribution
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("QUANTITY DISTRIBUTION")
print("=" * 60)

plt.figure(figsize=(8, 5))

plt.hist(
    df["Quantity"],
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Quantity per Order")
plt.xlabel("Quantity")
plt.ylabel("Number of Orders")
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 11. Outlier Analysis
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("OUTLIER ANALYSIS")
print("=" * 60)

# Calculate the IQR
Q1 = df["TotalPrice"].quantile(0.25)
Q3 = df["TotalPrice"].quantile(0.75)

IQR = Q3 - Q1

# Calculate acceptable limits
lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

# Identify outliers
outliers = df[
    (df["TotalPrice"] < lower_limit) |
    (df["TotalPrice"] > upper_limit)
]

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("\nLower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nNumber of Outliers:", len(outliers))

print("\nOutlier Records:")
print(
    outliers[
        ["OrderID", "Date", "Product", "Quantity", "TotalPrice"]
    ]
)


# ------------------------------------------------------------
# 12. Box Plot
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.boxplot(df["TotalPrice"])

plt.title("Box Plot of Total Price")
plt.ylabel("Total Price")
plt.grid(True)

plt.show()


# ------------------------------------------------------------
# 13. Key Observations
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("KEY OBSERVATIONS")
print("=" * 60)

most_ordered_product = (
    df["Product"]
    .value_counts()
    .idxmax()
)

highest_revenue_product = (
    df.groupby("Product")["TotalPrice"]
    .sum()
    .idxmax()
)

most_common_payment = (
    df["PaymentMethod"]
    .value_counts()
    .idxmax()
)

most_common_status = (
    df["OrderStatus"]
    .value_counts()
    .idxmax()
)

highest_revenue_month = monthly_revenue.idxmax()

print("\nAverage Order Value:",
      round(df["TotalPrice"].mean(), 2))

print("Median Order Value:",
      round(df["TotalPrice"].median(), 2))

print("Most Ordered Product:",
      most_ordered_product)

print("Highest Revenue Product:",
      highest_revenue_product)

print("Most Common Payment Method:",
      most_common_payment)

print("Most Common Order Status:",
      most_common_status)

print("Month with Highest Revenue:",
      highest_revenue_month)

print("Number of TotalPrice Outliers:",
      len(outliers))


print("\n" + "=" * 60)
print("PROJECT 2 ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 60)
```
